# Data and Experimental Protocol

This notebook shows how the released folds are organized and gives the basic statistics of the six train/test splits used in the paper.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Image

ROOT = Path('..').resolve()
RESULTS = ROOT / 'results'
FIGURES = ROOT / 'figures'
DATA = ROOT / 'data'
pd.set_option('display.max_columns', 80)
pd.set_option('display.max_colwidth', 120)
```

```python
folds = pd.read_csv(DATA / 'folds_summary.csv')
display(folds)
print('Total test table pairs:', folds['test_pair_id'].sum())
print('Total test rows:', folds['test_rows'].sum())
```

```python
# Load one fold as an example.
train0 = pd.read_parquet(DATA / 'folds/fold_0/train.parquet')
test0 = pd.read_parquet(DATA / 'folds/fold_0/test.parquet')
print(train0.shape, test0.shape)
display(train0.head())
```

The folds are built at table-pair level. This means that all candidate correspondences from the same table pair stay either in train or in test, never both.
