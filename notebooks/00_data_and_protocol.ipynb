# Data and Experimental Protocol

This notebook inspects the released folds and verifies that public fold files expose only the 60 MetaMatch meta-features used in the paper.

```
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
DATA = ROOT / 'data'
CONTEXT_COLUMNS = {
    'pair_id', 'dataset', 'relation_type', 'category',
    'source_table', 'target_table', 'source_column', 'target_column',
    'source_col_norm', 'target_col_norm', 'label',
}

def meta_feature_columns(df):
    return [c for c in df.columns if c not in CONTEXT_COLUMNS]
```

## Fold Summary

```
folds = pd.read_csv(DATA / 'folds_summary.csv')
display(folds)
print('Total test table pairs:', int(folds['test_pair_id'].sum()))
print('Total test rows:', int(folds['test_rows'].sum()))
```

## Released Columns

```
train0 = pd.read_parquet(DATA / 'folds/fold_0/train.parquet')
features = meta_feature_columns(train0)
print('fold_0 train shape:', train0.shape)
print('# context columns:', len([c for c in train0.columns if c in CONTEXT_COLUMNS]))
print('# paper meta-features:', len(features))
assert len(features) == 60
assert not any(c.startswith('spc_') or c.startswith('nlp_') or 'overlap' in c.lower() for c in features)
display(pd.DataFrame({'meta_feature': features}))
```

The split is built at table-pair level. All candidate correspondences from the same table pair stay either in train or in test, never both.
